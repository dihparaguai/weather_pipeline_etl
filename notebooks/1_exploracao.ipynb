# Extração e Transformação Inicial

```python
import json
import os
import pandas as pd
```

### 1. Capturar os dados extraídos

```python
dir_path = '../data/bronze'
# Lista nomes dos arquivos
lista_arquivos = [
    os.path.join(dir_path, f) 
    for f in os.listdir(dir_path) 
    if f.endswith('.json')
]

if not lista_arquivos:
    raise FileNotFoundError("Nenhum arquivo encontado na pasta data/. Rode o main.py primeiro.")

# Pega o último gerado caso tenha mais de um, ordenando pelo nome do arquivo (timestamp)
latest_file = max(lista_arquivos)
print(f"Carregando: {latest_file}")

with open(latest_file, 'r', encoding='utf-8') as f:
    bronze_data = json.load(f)
```

### 2. Criar um DF Pandas

```python
pd.set_option('display.max_columns', None)
df = pd.json_normalize(bronze_data)
df.head()
```
